```python
import numpy as np

def get_2polynomal_from_3_points(p0: np.array, p1: np.array, p2: np.array) -> np.array:
    # -- CODE --
    a = np.array([[p0[0]**2,p0[0],1.0], [p1[0]**2,p1[0],1.0], [p2[0]**2,p2[0],1.0]])
    print(a)
    b =np.array([p0[1],p1[1],p2[1]])
    print(b)

    return np.linalg.solve(a,b)
```
